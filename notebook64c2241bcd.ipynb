```python
DATASET_PATH = "/kaggle/input/datasets/suryach2/dataset"

MY_FACES_DIR = os.path.join(DATASET_PATH, "my_faces")
TEST_IMAGE_DIR = os.path.join(DATASET_PATH, "test_images")

MODEL_NAME = "ArcFace"
DETECTOR = "retinaface"

print("Dataset Path:", DATASET_PATH)
print("My Faces:", MY_FACES_DIR)
print("Test Images:", TEST_IMAGE_DIR)
```

```python
ext = {'.jpg', '.jpeg', '.png', '.bmp', '.webp'}

my_images = [
    f for f in Path(MY_FACES_DIR).iterdir()
    if f.suffix.lower() in ext
]

print(f"Found {len(my_images)} images")

for img in my_images:
    print(img.name)
```

```python
face_database = []

for img_path in my_images:
    try:
        print(f"Processing: {img_path.name}")

        embedding = DeepFace.represent(
            img_path=str(img_path),
            model_name=MODEL_NAME,
            detector_backend=DETECTOR,
            enforce_detection=False
        )

        face_database.append({
            "name": "Surya",
            "image": img_path.name,
            "embedding": embedding[0]["embedding"]
        })

    except Exception as e:
        print(f"Error processing {img_path.name}: {e}")
```

```python
PKL_PATH = "/kaggle/working/face_embeddings.pkl"

with open(PKL_PATH, "wb") as f:
    pickle.dump(face_database, f)

print("PKL file saved successfully")
print("Location:", PKL_PATH)
```

```python
with open(PKL_PATH, "rb") as f:
    saved_database = pickle.load(f)

print("Loaded embeddings:", len(saved_database))
print(saved_database[0].keys())
```

```python
from numpy.linalg import norm


def cosine_similarity(a, b):
    a = np.array(a)
    b = np.array(b)
    return np.dot(a, b) / (norm(a) * norm(b))
```

```python
TEST_IMAGE = os.path.join(TEST_IMAGE_DIR, "group_photo.jpg")

print(TEST_IMAGE)
```

```python
faces = DeepFace.extract_faces(
    img_path=TEST_IMAGE,
    detector_backend=DETECTOR,
    enforce_detection=False
)

print(f"Detected {len(faces)} faces")
```

```python
import matplotlib.pyplot as plt

THRESHOLD = 0.50

for i, face in enumerate(faces):

    temp_face_path = f"/kaggle/working/temp_face_{i}.jpg"

    # Convert face image
    face_img = (face["face"] * 255).astype("uint8")
    face_img = cv2.cvtColor(face_img, cv2.COLOR_RGB2BGR)

    cv2.imwrite(temp_face_path, face_img)

    # Get embedding
    query_embedding = DeepFace.represent(
        img_path=temp_face_path,
        model_name=MODEL_NAME,
        detector_backend="skip",
        enforce_detection=False
    )[0]["embedding"]

    best_score = -1
    best_name = "Unknown"

    # Compare with saved embeddings
    for person in saved_database:

        score = cosine_similarity(
            query_embedding,
            person["embedding"]
        )

        if score > best_score:
            best_score = score
            best_name = person["name"]

    # Final decision
    if best_score < THRESHOLD:
        best_name = "Unknown"

    # Print result
    print(f"\nFace {i+1}")
    print("Detected:", best_name)
    print("Similarity:", round(best_score, 4))

    # SHOW IMAGE
    plt.figure(figsize=(3,3))

    # Convert BGR to RGB for matplotlib
    rgb_face = cv2.cvtColor(face_img, cv2.COLOR_BGR2RGB)

    plt.imshow(rgb_face)
    plt.title(f"{best_name} ({best_score:.2f})")
    plt.axis("off")
    plt.show()
```
